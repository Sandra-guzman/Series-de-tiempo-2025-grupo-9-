# MODULO 4 : METODOLOGÍA BOX JENKINS.

INTRODUCCION:

El análisis de series de tiempo es fundamental para comprender y proyectar comportamientos de variables económicas y productivas a lo largo del tiempo. En este capítulo se aplicará la metodología Box-Jenkins para ajustar modelos ARIMA (Autoregresivos Integrados de Media Móvil) que permitan analizar la producción de café en Argentina.

La metodología Box-Jenkins (Box & Jenkins, 1970) propone un proceso iterativo que incluye la identificación, estimación, diagnóstico y predicción de modelos, garantizando que la serie sea estacionaria —es decir, que su media y varianza permanezcan estables en el tiempo—. Para seleccionar el modelo más adecuado se emplean criterios de información como el AIC y el BIC, que equilibran precisión y simplicidad del ajuste.

En el contexto agrícola, los modelos ARIMA facilitan la detección de patrones históricos y la proyección de escenarios futuros, lo que apoya la planeación agroindustrial y la gestión de riesgos productivos. Este capítulo busca demostrar la aplicabilidad de los modelos estacionarios como herramienta analítica para anticipar variaciones en la producción cafetera argentina.

CONCLUSIONES  GENERALES

El análisis de la serie temporal de producción de café en Argentina permitió identificar comportamientos irregulares y una marcada variabilidad interanual. La prueba ADF confirmó que la serie original no era estacionaria, por lo que se aplicó una diferenciación de primer orden, obteniendo una serie estacionaria y apta para modelación bajo la metodología Box–Jenkins. A partir de las funciones ACF y PACF, se identificó un patrón compatible con un modelo ARIMA(1,1,1), el cual ofreció un ajuste adecuado y métricas óptimas (AIC = 1123.04, BIC = 1127.62). Los residuos del modelo se comportaron como ruido blanco, sin autocorrelación significativa ni sesgos estructurales, validando la consistencia del ajuste.

El pronóstico mostró una tendencia estable en los próximos años (2025–2029), con niveles de producción cercanos a 4.5 millones de toneladas, lo que sugiere un periodo de relativa estabilidad sin variaciones abruptas en el corto plazo. No obstante, la detección de puntos de cambio reveló una alteración estructural en la media, posiblemente asociada a transformaciones económicas, tecnológicas o climáticas que incidieron en el rendimiento agrícola. Por otro lado, el análisis de valores atípicos no evidenció observaciones fuera de los límites de control (±3σ), lo que confirma la ausencia de perturbaciones extremas y respalda la robustez del modelo.

En conjunto, los resultados demuestran que el modelo ARIMA(1,1,1) describe adecuadamente la dinámica de la producción cafetera argentina y proporciona una herramienta confiable para la proyección de tendencias. Se recomienda, sin embargo, explorar versiones extendidas como SARIMA, que incorporen posibles efectos estacionales, o Prophet, para evaluar escenarios con mayor flexibilidad ante fluctuaciones irregulares. Este enfoque integral coincide con las recomendaciones metodológicas de Box, Jenkins, Reinsel y Ljung (2015) y Hyndman y Athanasopoulos (2018), quienes destacan la importancia de combinar rigor estadístico con interpretación contextual en la modelación de series temporales agrícolas.
